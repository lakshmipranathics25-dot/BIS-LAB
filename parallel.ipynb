```python

from concurrent.futures import ThreadPoolExecutor

intersections = [
    {"name": "I1", "NS": 30, "EW": 10},
    {"name": "I2", "NS": 15, "EW": 25},
    {"name": "I3", "NS": 40, "EW": 20},
    {"name": "I4", "NS": 10, "EW": 30}
]

CYCLE = 80
MIN_GREEN = 20
MAX_GREEN = 60

def update_cell(i, old_times):

    ns = intersections[i]["NS"]
    ew = intersections[i]["EW"]

    neighbors = [j for j in (i-1, i+1)
                 if 0 <= j < len(intersections)]

    ns += 0.2 * sum(intersections[j]["NS"] for j in neighbors)
    ew += 0.2 * sum(intersections[j]["EW"] for j in neighbors)

    target = round(CYCLE * ns / (ns + ew))

    target = max(MIN_GREEN, min(MAX_GREEN, target))

    new_ns = round(0.5 * old_times[i][0] + 0.5 * target)
    new_ew = CYCLE - new_ns

    return (new_ns, new_ew)


times = [(40, 40)] * len(intersections)

for iteration in range(3):

    with ThreadPoolExecutor(max_workers=4) as executor:
        times = list(executor.map(
            lambda i: update_cell(i, times),
            range(len(intersections))
        ))

# Display final results
print("OPTIMIZED TRAFFIC SIGNAL TIMINGS")
print("--------------------------------")

for i in range(len(intersections)):
    print(intersections[i]["name"])
    print("North-South Green:", times[i][0], "seconds")
    print("East-West Green:", times[i][1], "seconds")
    print()

print("Traffic signal optimization completed!")
```
